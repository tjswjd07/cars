```python
# 모델 및 fuel타입으로 구하기 -> 동일 모델, 연료종류 없음 
# 보고 싶은 조합 (모델명 + 연료 종류)
target_combos = [
    ('TOYOTA MARK X', 'Petrol'),
    ('NISSAN CUBIC', 'Petrol'),
    ('ISUZU TROOPER', 'Diesel'),
    ('VAUXHALL AMPERA', 'Petrol Plug-in Hybrid'),
]

# 각 조합에 대해 최빈값 출력
for model, fuel in target_combos:
    # 조건에 맞는 행만 필터링
    data = df_cars[(df_cars['title'] == model) & 
                   (df_cars['Fuel type'] == fuel) & 
                   (df_cars['Engine'].notna())]
    
    # 최빈값(mode) 출력
    if not data.empty:
        mode_val = data['Engine']().iloc[0]
        print(f"{model} / {fuel} → Engine 최빈값: {mode_val}")
    else:
        print(f"{model} / {fuel} → 데이터 없음 (mode 구할 수 없음)")

# 모델로만 구하기 -> 해당 모델과 동일 모델 없음
# 보고 싶은 모델 리스트
target_models = ['TOYOTA MARK X', 'NISSAN CUBIC', 'ISUZU TROOPER', 'VAUXHALL AMPERA']

for model in target_models:
    # 해당 모델의 엔진 정보 (결측값 제외)
    data = df_cars[(df_cars['title'] == model) & (df_cars['Engine'].notna())]
    
    # 최빈값 구하기
    if not data.empty:
        mode_val = data['Engine'].mode().iloc[0]
        print(f"{model} → Engine 최빈값: {mode_val}")
    else:
        print(f"{model} → 데이터 없음 (mode 구할 수 없음)")



#추측할 수 있는 base데이터 없으므로 삭제 (실무에서라면 외부에서 데이터를 가지고 오겠지만.. 삭제 하겠습니다.)
```

```python
#engine에서 남은 결측치 삭제
df_cars = df_cars[~df_cars['Engine'].isna()]
```

```python
#(3)Emission Class 결측치 처리하기  
emission_miss = df_cars[df_cars['Emission Class'].isna()]
emission_miss.info() # 56개 결측치 
emission_miss
```

```python
#(2) Emission class 결측값을 연도와 fuel type기준 mode값으로 채워넣기
# 1. 연도 그룹핑 함수 정의
def get_year_group(year):
    if pd.isna(year):
        return None
    elif year <= 2000:
        return 'before_2000'
    elif year <= 2005:
        return '2001_2005'
    elif year <= 2010:
        return '2006_2010'
    elif year <= 2014:
        return '2011_2014'
    else:
        return 'after_2015'

# 2. 연도 그룹 열 생성
df_cars['year_group'] = df_cars['Registration_Year'].apply(get_year_group)

# 3. 연도 그룹 + 연료 유형 기준으로 Emission Class의 mode 계산
emission_group_mode = (
    df_cars[df_cars['Emission Class'].notna()]
    .groupby(['year_group', 'Fuel type'])['Emission Class']
    .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else None)
    .reset_index()
    .rename(columns={'Emission Class': 'Emission_mode'})
)

# 4. 원본 데이터와 mode 테이블 merge
df_cars = df_cars.merge(emission_group_mode, on=['year_group', 'Fuel type'], how='left')

# 5. Emission Class가 결측인 경우에만 mode값으로 채우기
df_cars.loc[df_cars['Emission Class'].isna(), 'Emission Class'] = df_cars.loc[df_cars['Emission Class'].isna(), 'Emission_mode']

# 6. 보조 열 제거
df_cars.drop(columns=['Emission_mode', 'year_group'], inplace=True)

# 7. 결측치 확인
print("✅ 남은 Emission Class 결측치 수:", df_cars['Emission Class'].isna().sum())
```

```python
#STEP4. 범주형 데이터, 문자열 수정 
#(1) cars register year > 연식으로 계산
#(2) Engine 'L'표기 삭제
#(3) Emission class에서 Euro 빼기
#(4) title 에서 제조사 추출
#(5) Service History NaN은 unkown으로 바꾸기
#(6) 범주형 데이터 숫자형으로 변경하기
```

```python
from datetime import datetime

# (1) Registration_Year → Car_Age
current_year = datetime.now().year
df_cars['Car_Age'] = current_year - df_cars['Registration_Year']

# (2) Engine: "1.4L" → 1.4
df_cars['Engine'] = df_cars['Engine'].str.replace('L', '', regex=False).astype(float)

# (3) Emission Class: "Euro 6" → 6
df_cars['Num_Emission_Class'] = df_cars['Emission Class'].str.extract('(\d+)').astype(float)

# (4) Brand 추출: "SKODA FABIA" → "SKODA"
df_cars['Brand'] = df_cars['title'].str.split().str[0]


df_cars.head()
```

```python
#(5) Service History NaN은 unkown으로 바꾸기
df_cars['Service history'] = df_cars['Service history'].fillna('Unknown')
```

```python
#(6) 범주형 데이터 숫자형으로 변경하기
df_cars.info()   #범주형 데이터 확인하기 Fuel type, Body type, Gearbox, Emission Class(Num_Emission_Class로 변경), Service history

#범주형 데이터, 고유값 갯수 확인하기
df_cars['Fuel type'].value_counts() # 고유값 : Petrol, Diesel, Petrol Hybrid, Petrol Plug-in Hybrid, Electric, Desel hybrid 
df_cars['Body type'].value_counts() #고유값 : Hatchback, SUV, Saloon, MPV, Estate, Coupe, Convertible, Pickup, Combi Van 
df_cars['Service history'].value_counts() #고유값 : Unkown, Full

#고유값의 갯수가 적은 것 (1%미만 기준) others로 묶기
# Fuel Type
df_cars['Fuel type'] = df_cars['Fuel type'].replace({
    'Petrol Hybrid': 'Others',
    'Petrol Plug-in Hybrid': 'Others',
    'Electric': 'Others',
    'Diesel Hybrid': 'Others'
})

# Body type
df_cars['Body type'] = df_cars['Body type'].replace({
    'Pickup': 'Other',
    'Combi Van': 'Other'
})

#원-핫코딩 *prefix를 활용해 열이름 구분
df_cars = pd.get_dummies(df_cars, columns=['Fuel type', 'Body type', 'Service history'],
               prefix=['Fuel', 'Body', 'Service'], dtype=int, drop_first=True)
```

```python
#STEP1. 스캐일러를 활용해서 데이터 표준화 
#정규분포를 따르지 않으므로 Robust Scaler활용
!pip install scikit-learn
from sklearn.preprocessing import RobustScaler

df_scaled = RobustScaler().fit_transform(df_cars.select_dtypes(include='number'))  #숫자형 열만 진행
df_cars[df_cars.select_dtypes(include='number').columns] = df_scaled #기존 데이터 프레임에 붙여넣기
```

```python
#STEP2. 주성분분석(PCA) -> 전체 분산의 80%이상 설명되도록 주성분 갯수 선택하기
from sklearn.decomposition import PCA
pca = PCA(n_components=0.80)
df_pca = pca.fit_transform(df_scaled)
```

```python
print(f"선택된 주성분 수: {pca.n_components_}")
```

```python
print(f"누적 설명력: {pca.explained_variance_ratio_.sum():.4f}")
#선택된 주성분 4개로, 전체 분산의 80% 설명
```

```python
#PCA그래프 그리기
import matplotlib.pyplot as plt
import pandas as pd

# PCA 결과를 DataFrame으로 변환 (2개 주성분만 시각화)
df_pca_df = pd.DataFrame(df_pca, columns=[f'PC{i+1}' for i in range(df_pca.shape[1])])

# 2D Scatter Plot
plt.figure(figsize=(8,6))
plt.scatter(df_pca_df['PC1'], df_pca_df['PC2'], alpha=0.5)
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('PCA 2D Scatter Plot')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# df_scaled: 스케일된 원 데이터
# pca: 학습된 PCA 객체
# df_pca: PCA 적용 결과

plt.figure(figsize=(10, 8))
# 1. 샘플 좌표 (PC1, PC2)
plt.scatter(df_pca[:, 0], df_pca[:, 1], alpha=0.3)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Bi-Plot")

# 2. 변수 벡터 (각 변수의 주성분 기여도 방향)
features = df_cars.select_dtypes(include='number').columns
for i, feature in enumerate(features):
    plt.arrow(0, 0,  # 시작점
              pca.components_[0, i]*5,  # PC1 축 기여
              pca.components_[1, i]*5,  # PC2 축 기여
              color='red', alpha=0.5, head_width=0.1)
    plt.text(pca.components_[0, i]*5.2, pca.components_[1, i]*5.2, 
             feature, color='green', ha='center', va='center')

plt.grid()
plt.show()
```

```python
import pandas as pd
import numpy as np

# 1. 상관계수 행렬 계산 (수치형 변수만)
corr_matrix = df_cars.corr(numeric_only=True)

# 2. 상삼각 행렬 마스크 적용해서 중복 제거 (자기자신 포함 안함)
mask = np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
corr_pairs = corr_matrix.where(mask)

# 3. stack으로 쌍 변환
corr_pairs = corr_pairs.stack().reset_index()
corr_pairs.columns = ['Variable 1', 'Variable 2', 'Correlation']

# 4. 절댓값 기준 상관계수가 높은 순으로 정렬
corr_pairs['Abs Correlation'] = corr_pairs['Correlation'].abs()
corr_pairs = corr_pairs.sort_values(by='Abs Correlation', ascending=False)

# 5. 상위 50개 출력
top_50_corr = corr_pairs.head(50).drop(columns='Abs Correlation')
print(top_50_corr)
```
